```python
from pathlib import Path
import json

forecast_path = Path("latest_forecast_data.json")
with forecast_path.open("r", encoding="utf-8") as f:
    forecast_data = json.load(f)

array_size = len(forecast_data) if hasattr(forecast_data, "__len__") else None

variable_names = None
if isinstance(forecast_data, dict):
    variable_names = list(forecast_data.keys())
elif isinstance(forecast_data, list) and forecast_data and isinstance(forecast_data[0], dict):
    variable_names = list(forecast_data[0].keys())

print(f"Array size: {array_size}")
print(f"Variable names: {variable_names}")

forecast_data
```
